```python
import os
from datasets import load_dataset
from transformers import AutoTokenizer, DataCollatorWithPadding

MODEL_NAME = "distilbert-base-uncased"
NUM_LABELS = 4
LABEL_NAMES = ["World", "Sports", "Business", "Sci/Tech"]

# Загружаем датасет AG News напрямую через его Parquet-версию на Hugging Face, минуя устаревшие загрузочные скрипты
raw_ds = load_dataset("fancyzhx/ag_news")

# Tez va muvozanatli: 2500 train, 500 test
train_ds = raw_ds["train"].shuffle(seed=SEED).select(range(2500))
test_ds = raw_ds["test"].shuffle(seed=SEED).select(range(500))

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_fn(batch):
    return tokenizer(batch["text"], truncation=True, max_length=80)

train_tok = train_ds.map(tokenize_fn, batched=True, remove_columns=["text"])
test_tok = test_ds.map(tokenize_fn, batched=True, remove_columns=["text"])
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

def get_dir_size_mb(path):
    return sum(
        os.path.getsize(os.path.join(r, f))
        for r, _, fs in os.walk(path)
        for f in fs
    ) / (1024 * 1024)
```
