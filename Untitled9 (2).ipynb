```python
# 1-topshiriq: Titanic, Keras Functional (Embedding + Normalization), fit vs GradientTape

# %% 1. Kerakli kutubxonalar
import time
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split

# Sozlamalar: seed, epoch soni, batch hajmi, learning rate
SEED, EPOCHS, BATCH, LR = 42, 20, 32, 1e-3

# Seed qo'yamiz: har safar ishlaganda natija bir xil chiqadi
keras.utils.set_random_seed(SEED)
tf.config.experimental.enable_op_determinism()

# %% 2. Datasetni yuklaymiz
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)

# Kabina raqamidan faqat birinchi harfni olamiz (paluba), bo'sh bo'lsa "U"
df["Deck"] = df["Cabin"].fillna("U").str[0]
# Bo'sh qolgan port nomini "NA" bilan to'ldiramiz
df["Embarked"] = df["Embarked"].fillna("NA")
# Pclass raqam, lekin aslida kategoriya, shuning uchun matnga o'giramiz
df["Pclass"] = df["Pclass"].astype(str)

CAT = ["Pclass", "Sex", "Embarked", "Deck"]   # kategorik ustunlar
NUM = ["Age", "SibSp", "Parch", "Fare"]       # sonli ustunlar

# Train (80%) va validation (20%) ga ajratamiz
train_df, val_df = train_test_split(
    df, test_size=0.2, random_state=SEED, stratify=df["Survived"])
train_df, val_df = train_df.copy(), val_df.copy()

# Yoshdagi bo'sh joylarni train o'rtachasi (median) bilan to'ldiramiz
age_med = train_df["Age"].median()
for d in (train_df, val_df):
    d["Age"] = d["Age"].fillna(age_med)


def to_xy(d):
    # Kategorik ustunlarni matn, sonlilarni float qilib, X va y ga ajratamiz
    x = {c: d[c].astype(str).to_numpy().reshape(-1, 1) for c in CAT}
    x["num"] = d[NUM].to_numpy("float32")
    return x, d["Survived"].to_numpy("float32")


x_tr, y_tr = to_xy(train_df)
x_va, y_va = to_xy(val_df)


def make_ds(x, y, optimized):
    # Batchlarga bo'lamiz. Shuffle yo'q, shuning uchun tartib har doim bir xil
    ds = tf.data.Dataset.from_tensor_slices((x, y)).batch(BATCH)
    if optimized:
        # cache: ma'lumotni xotirada saqlaydi
        # prefetch: keyingi batchni oldindan tayyorlab qo'yadi
        ds = ds.cache().prefetch(tf.data.AUTOTUNE)
    return ds


# %% 3. Qayta ishlash qatlamlari (faqat train ma'lumotidan o'rganadi)
lookups = {}
for c in CAT:
    lk = layers.StringLookup()      # matnni raqamga (indeksga) aylantiradi
    lk.adapt(x_tr[c])
    lookups[c] = lk

norm = layers.Normalization()       # sonlarni o'rtacha 0, og'ish 1 qiladi
norm.adapt(x_tr["num"])


# %% 4. Modelni quramiz (ikki tarmoqli Functional API)
def build_model():
    # Har safar bir xil seed: ikkala modelning boshlang'ich og'irliklari bir xil
    keras.utils.set_random_seed(SEED)

    # --- 1-tarmoq: kategorik ustunlar ---
    inputs, cat_vecs = {}, []
    for c in CAT:
        inp = keras.Input(shape=(1,), dtype="string", name=c)
        inputs[c] = inp
        idx = lookups[c](inp)                                         # matn -> indeks
        emb = layers.Embedding(lookups[c].vocabulary_size(), 4)(idx)  # indeks -> 4 ta son
        cat_vecs.append(layers.Flatten()(emb))
    cat_branch = layers.Dense(16, activation="relu")(layers.Concatenate()(cat_vecs))

    # --- 2-tarmoq: sonli ustunlar ---
    num_in = keras.Input(shape=(len(NUM),), name="num")
    inputs["num"] = num_in
    num_branch = layers.Dense(16, activation="relu")(norm(num_in))    # normallash + Dense

    # --- Ikkalasini birlashtiramiz ---
    h = layers.Concatenate()([cat_branch, num_branch])
    h = layers.Dense(16, activation="relu")(h)
    out = layers.Dense(1, activation="sigmoid")(h)   # 0 yoki 1 ehtimoli (omon qoldimi)

    m = keras.Model(inputs, out)
    m.compile(optimizer=keras.optimizers.Adam(LR),
              loss="binary_crossentropy", metrics=["accuracy"])
    return m


# %% 5. Ikki xil o'qitish usuli
def train_fit(optimized=True):
    # 1-usul: tayyor model.fit
    ds = make_ds(x_tr, y_tr, optimized)
    m = build_model()
    t = time.perf_counter()
    m.fit(ds, epochs=EPOCHS, verbose=0)
    return m, time.perf_counter() - t


def train_tape(optimized=True):
    # 2-usul: siklni qo'lda yozamiz (GradientTape)
    ds = make_ds(x_tr, y_tr, optimized)
    m = build_model()
    m.optimizer.build(m.trainable_variables)
    loss_fn = keras.losses.BinaryCrossentropy()

    @tf.function    # kodni tezlashtiradi
    def step(x, y):
        with tf.GradientTape() as tape:
            p = m(x, training=True)                      # 1) bashorat
            loss = loss_fn(tf.expand_dims(y, -1), p)     # 2) xatoni hisoblash
        grads = tape.gradient(loss, m.trainable_variables)             # 3) gradient
        m.optimizer.apply_gradients(zip(grads, m.trainable_variables)) # 4) og'irlikni yangilash
        return loss

    t = time.perf_counter()
    for _ in range(EPOCHS):          # har epoch
        for x, y in ds:              # har batch
            step(x, y)
    return m, time.perf_counter() - t


val_ds = make_ds(x_va, y_va, True)

# Ikkalasini o'qitamiz
m_fit, t_fit = train_fit(True)
m_tape, t_tape = train_tape(True)

# Validation'da tekshiramiz
loss_fit, acc_fit = m_fit.evaluate(val_ds, verbose=0)
loss_tape, acc_tape = m_tape.evaluate(val_ds, verbose=0)

# Ikki modelning bashoratlari qanchalik bir xil ekanini tekshiramiz
max_diff = np.abs(m_fit.predict(val_ds, verbose=0) -
                  m_tape.predict(val_ds, verbose=0)).max()

# Natija jadvali
res = pd.DataFrame({
    "Usul": ["model.fit", "GradientTape"],
    "Val accuracy": [acc_fit, acc_tape],
    "Val loss": [loss_fit, loss_tape],
    "Training time (s)": [t_fit, t_tape],
})
print(res.round(4).to_string(index=False))
print("Bashoratlar orasidagi maksimal farq:", float(max_diff))   # 0 ga yaqin bo'lishi kerak


# %% 6. tf.data tezligi: cache().prefetch() bilan va usiz
def bench_iter(optimized, epochs=50):
    # Faqat ma'lumotni aylanib chiqish vaqtini o'lchaymiz
    ds = make_ds(x_tr, y_tr, optimized)
    t = time.perf_counter()
    for _ in range(epochs):
        for _ in ds:
            pass
    return time.perf_counter() - t


bench_iter(False, 2); bench_iter(True, 2)   # isitib olish (o'lchovga kirmaydi)

# To'liq o'qitish vaqti ikkala variantda
_, t_fit_plain = train_fit(False)
_, t_fit_opt = train_fit(True)

pipe = pd.DataFrame({
    "Variant": ["Optimallashtirilmagan", "cache().prefetch()"],
    "Faqat data iteratsiyasi, 50 epoch (s)": [bench_iter(False), bench_iter(True)],
    f"model.fit, {EPOCHS} epoch (s)": [t_fit_plain, t_fit_opt],
})
print(pipe.round(4).to_string(index=False))
```
