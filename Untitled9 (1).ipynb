```python
# 1-topshiriq: Titanic, Keras Functional (Embedding + Normalization), fit vs GradientTape
# Google Colab: har "# %%" blokni alohida katakka qo'ying (yoki hammasini bitta katakka).

# %% 1. Kutubxonalar va seed (reproducibility)
import time
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split

SEED, EPOCHS, BATCH, LR = 42, 20, 32, 1e-3
keras.utils.set_random_seed(SEED)               # python, numpy, tf seedlari
tf.config.experimental.enable_op_determinism()  # deterministik operatsiyalar

# %% 2. Ma'lumot
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)
df["Deck"] = df["Cabin"].fillna("U").str[0]
df["Embarked"] = df["Embarked"].fillna("NA")
df["Pclass"] = df["Pclass"].astype(str)

CAT = ["Pclass", "Sex", "Embarked", "Deck"]
NUM = ["Age", "SibSp", "Parch", "Fare"]

train_df, val_df = train_test_split(
    df, test_size=0.2, random_state=SEED, stratify=df["Survived"])
train_df, val_df = train_df.copy(), val_df.copy()
age_med = train_df["Age"].median()              # faqat train'dan (leakage yo'q)
for d in (train_df, val_df):
    d["Age"] = d["Age"].fillna(age_med)


def to_xy(d):
    x = {c: d[c].astype(str).to_numpy().reshape(-1, 1) for c in CAT}
    x["num"] = d[NUM].to_numpy("float32")
    return x, d["Survived"].to_numpy("float32")


x_tr, y_tr = to_xy(train_df)
x_va, y_va = to_xy(val_df)


def make_ds(x, y, optimized):
    # Shuffle YO'Q: ikkala usul aynan bir xil batch tartibini ko'radi
    ds = tf.data.Dataset.from_tensor_slices((x, y)).batch(BATCH)
    if optimized:
        ds = ds.cache().prefetch(tf.data.AUTOTUNE)
    return ds


# %% 3. Preprocessing qatlamlari (trainable emas, ikkala modelda umumiy bo'lishi mumkin)
lookups = {}
for c in CAT:
    lk = layers.StringLookup()
    lk.adapt(x_tr[c])
    lookups[c] = lk
norm = layers.Normalization()
norm.adapt(x_tr["num"])


# %% 4. Ikki tarmoqli Functional model
def build_model():
    keras.utils.set_random_seed(SEED)           # har safar bir xil boshlang'ich og'irliklar
    inputs, cat_vecs = {}, []
    for c in CAT:
        inp = keras.Input(shape=(1,), dtype="string", name=c)
        inputs[c] = inp
        idx = lookups[c](inp)
        emb = layers.Embedding(lookups[c].vocabulary_size(), 4)(idx)
        cat_vecs.append(layers.Flatten()(emb))
    cat_branch = layers.Dense(16, activation="relu")(layers.Concatenate()(cat_vecs))

    num_in = keras.Input(shape=(len(NUM),), name="num")
    inputs["num"] = num_in
    num_branch = layers.Dense(16, activation="relu")(norm(num_in))

    h = layers.Concatenate()([cat_branch, num_branch])
    h = layers.Dense(16, activation="relu")(h)
    out = layers.Dense(1, activation="sigmoid")(h)

    m = keras.Model(inputs, out)
    m.compile(optimizer=keras.optimizers.Adam(LR),
              loss="binary_crossentropy", metrics=["accuracy"])
    return m


# %% 5. Ikki o'qitish usuli
def train_fit(optimized=True):
    ds = make_ds(x_tr, y_tr, optimized)
    m = build_model()
    t = time.perf_counter()
    m.fit(ds, epochs=EPOCHS, verbose=0)
    return m, time.perf_counter() - t


def train_tape(optimized=True):
    ds = make_ds(x_tr, y_tr, optimized)
    m = build_model()
    m.optimizer.build(m.trainable_variables)
    loss_fn = keras.losses.BinaryCrossentropy()

    @tf.function
    def step(x, y):
        with tf.GradientTape() as tape:
            p = m(x, training=True)
            loss = loss_fn(tf.expand_dims(y, -1), p)
        grads = tape.gradient(loss, m.trainable_variables)
        m.optimizer.apply_gradients(zip(grads, m.trainable_variables))
        return loss

    t = time.perf_counter()
    for _ in range(EPOCHS):
        for x, y in ds:
            step(x, y)
    return m, time.perf_counter() - t


val_ds = make_ds(x_va, y_va, True)

m_fit, t_fit = train_fit(True)
m_tape, t_tape = train_tape(True)

loss_fit, acc_fit = m_fit.evaluate(val_ds, verbose=0)
loss_tape, acc_tape = m_tape.evaluate(val_ds, verbose=0)
max_diff = np.abs(m_fit.predict(val_ds, verbose=0) -
                  m_tape.predict(val_ds, verbose=0)).max()

res = pd.DataFrame({
    "Usul": ["model.fit", "GradientTape"],
    "Val accuracy": [acc_fit, acc_tape],
    "Val loss": [loss_fit, loss_tape],
    "Training time (s)": [t_fit, t_tape],
})
print(res.round(4).to_string(index=False))
print("Bashoratlar orasidagi maksimal farq:", float(max_diff))

# %% 6. tf.data: cache().prefetch() bilan va usiz
def bench_iter(optimized, epochs=50):
    ds = make_ds(x_tr, y_tr, optimized)
    t = time.perf_counter()
    for _ in range(epochs):
        for _ in ds:
            pass
    return time.perf_counter() - t


bench_iter(False, 2); bench_iter(True, 2)       # warm-up
_, t_fit_plain = train_fit(False)
_, t_fit_opt = train_fit(True)

pipe = pd.DataFrame({
    "Variant": ["Optimallashtirilmagan", "cache().prefetch()"],
    "Faqat data iteratsiyasi, 50 epoch (s)": [bench_iter(False), bench_iter(True)],
    f"model.fit, {EPOCHS} epoch (s)": [t_fit_plain, t_fit_opt],
})
print(pipe.round(4).to_string(index=False))
```
