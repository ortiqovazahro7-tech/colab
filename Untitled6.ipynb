```python
!pip install -q importlib_resources

import gc
import time
import numpy as np
import pandas as pd
import tensorflow as tf
import tensorflow_datasets as tfds
from sklearn.model_selection import train_test_split
from tensorflow import keras
from tensorflow.keras import layers

SEED, IMG, BATCH, EPOCHS, PER_CLASS = 42, 224, 32, 5, 400
AUTOTUNE = tf.data.AUTOTUNE
keras.utils.set_random_seed(SEED)

raw = tfds.load("tf_flowers", split="train", as_supervised=True)
raw = raw.map(lambda x, y: (tf.cast(tf.image.resize(x, (IMG, IMG)), tf.uint8), y),
              num_parallel_calls=AUTOTUNE).batch(256)
X_all = np.concatenate([x.numpy() for x, _ in raw])
y_all = np.concatenate([y.numpy() for _, y in raw])
NUM_CLASSES = int(y_all.max()) + 1

rng = np.random.default_rng(SEED)
chosen = np.concatenate([
    rng.choice(np.where(y_all == c)[0], PER_CLASS, replace=False) for c in range(NUM_CLASSES)])
X, y = X_all[chosen], y_all[chosen]
del X_all, y_all
gc.collect()

X_tr, X_tmp, y_tr, y_tmp = train_test_split(X, y, test_size=0.30, stratify=y, random_state=SEED)
X_va, X_te, y_va, y_te = train_test_split(X_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=SEED)
N_TEST = len(X_te)
print("Sinflar:", NUM_CLASSES, "| sinfiga:", PER_CLASS, "| train/val/test:", len(X_tr), len(X_va), N_TEST)


def make_ds(Xa, ya, fn, train):
    ds = tf.data.Dataset.from_tensor_slices((Xa, ya))
    ds = ds.map(lambda a, b: (fn(tf.cast(a, tf.float32)), b), num_parallel_calls=AUTOTUNE)
    if train:
        ds = ds.shuffle(len(Xa), seed=SEED)
    return ds.batch(BATCH).prefetch(AUTOTUNE)


BASES = {
    "MobileNetV2": (keras.applications.MobileNetV2,
                    keras.applications.mobilenet_v2.preprocess_input),
    "EfficientNetB0": (keras.applications.EfficientNetB0,
                       keras.applications.efficientnet.preprocess_input),
    "ResNet50": (keras.applications.ResNet50,
                 keras.applications.resnet50.preprocess_input),
}

STAGES = [
    ("1) feature extraction", "frozen", 1e-3),
    ("2) oxirgi 30% ochiq", "last30", 1e-4),
    ("3) butun baza (lr 1e-5)", "all", 1e-5),
]


def set_trainable(base, mode):
    if mode == "frozen":
        base.trainable = False
        return
    base.trainable = True
    for l in base.layers:
        if isinstance(l, layers.BatchNormalization):
            l.trainable = False
    if mode == "last30":
        k = int(len(base.layers) * 0.7)
        for l in base.layers[:k]:
            l.trainable = False


def count_trainable(m):
    return int(sum(np.prod(w.shape) for w in m.trainable_weights))


def reset_gpu():
    try:
        tf.config.experimental.reset_memory_stats("GPU:0")
    except Exception:
        pass


def gpu_peak_mb():
    try:
        return tf.config.experimental.get_memory_info("GPU:0")["peak"] / 1024 ** 2
    except Exception:
        return np.nan


def inference_ms(m, ds):
    m.predict(ds.take(1), verbose=0)
    t = time.perf_counter()
    m.predict(ds, verbose=0)
    return (time.perf_counter() - t) / N_TEST * 1000


rows = []
for bname, (ctor, pre) in BASES.items():
    keras.backend.clear_session()
    gc.collect()
    keras.utils.set_random_seed(SEED)

    tr = make_ds(X_tr, y_tr, pre, True)
    va = make_ds(X_va, y_va, pre, False)
    te = make_ds(X_te, y_te, pre, False)

    base = ctor(include_top=False, weights="imagenet", input_shape=(IMG, IMG, 3))
    inp = keras.Input(shape=(IMG, IMG, 3))
    x = base(inp, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.2)(x)
    out = layers.Dense(NUM_CLASSES, activation="softmax")(x)
    model = keras.Model(inp, out)

    for sname, mode, lr in STAGES:
        set_trainable(base, mode)
        model.compile(optimizer=keras.optimizers.Adam(lr),
                      loss="sparse_categorical_crossentropy", metrics=["accuracy"])
        reset_gpu()
        t = time.perf_counter()
        model.fit(tr, validation_data=va, epochs=EPOCHS, verbose=0)
        train_time = time.perf_counter() - t
        peak = gpu_peak_mb()
        acc = model.evaluate(te, verbose=0)[1]
        rows.append({
            "Baza": bname,
            "Strategiya": sname,
            "Test accuracy": acc,
            "Trainable params": count_trainable(model),
            "Train vaqti (s)": train_time,
            "Inference (ms/rasm)": inference_ms(model, te),
            "GPU peak (MB)": peak,
        })
        print(bname, sname, "tayyor")

res = pd.DataFrame(rows)
print(res.round(4).to_string(index=False))

best_acc = res.loc[res["Test accuracy"].idxmax()]
fastest = res.loc[res["Inference (ms/rasm)"].idxmin()]
print("\nEng yuqori accuracy:", best_acc["Baza"], "|", best_acc["Strategiya"],
      "|", round(best_acc["Test accuracy"], 4))
print("Eng tez inference:", fastest["Baza"], "|", fastest["Strategiya"],
      "|", round(fastest["Inference (ms/rasm)"], 3), "ms")
```
