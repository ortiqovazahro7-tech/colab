```python
!pip install -q peft datasets

import gc
import os
import time
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
from datasets import load_dataset
from transformers import (AutoTokenizer, AutoModelForSequenceClassification,
                          TrainingArguments, Trainer, DataCollatorWithPadding, set_seed)
from peft import LoraConfig, get_peft_model, TaskType
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, ConfusionMatrixDisplay

MODEL, SEED, MAXLEN, EPOCHS = "distilbert-base-uncased", 42, 64, 3
N_TRAIN, N_VAL, N_TEST = 8000, 1000, 3000
LABELS = ["negative", "neutral", "positive"]
set_seed(SEED)

ds = load_dataset("cardiffnlp/tweet_eval", "sentiment")
train = ds["train"].shuffle(seed=SEED).select(range(N_TRAIN))
val = ds["validation"].shuffle(seed=SEED).select(range(N_VAL))
test = ds["test"].shuffle(seed=SEED).select(range(N_TEST))
test_texts = test["text"]
test_labels = np.array(test["label"])
print("Train/val/test:", len(train), len(val), len(test))

tok = AutoTokenizer.from_pretrained(MODEL)


def tokenize(batch):
    return tok(batch["text"], truncation=True, max_length=MAXLEN)


train_tok = train.map(tokenize, batched=True)
val_tok = val.map(tokenize, batched=True)
test_tok = test.map(tokenize, batched=True)
collator = DataCollatorWithPadding(tok)


def compute_metrics(eval_pred):
    logits, labels = eval_pred
    pred = np.argmax(logits, axis=-1)
    return {"accuracy": accuracy_score(labels, pred),
            "f1": f1_score(labels, pred, average="macro")}


def build(method, r=None):
    set_seed(SEED)
    model = AutoModelForSequenceClassification.from_pretrained(MODEL, num_labels=len(LABELS))
    if method == "head":
        for p in model.distilbert.parameters():
            p.requires_grad = False
    elif method == "lora":
        cfg = LoraConfig(task_type=TaskType.SEQ_CLS, r=r, lora_alpha=2 * r,
                         lora_dropout=0.1, target_modules=["q_lin", "v_lin"])
        model = get_peft_model(model, cfg)
    return model


def count_trainable(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)


def count_total(model):
    return sum(p.numel() for p in model.parameters())


def dir_size_mb(path):
    total = 0
    for root, _, files in os.walk(path):
        for f in files:
            total += os.path.getsize(os.path.join(root, f))
    return total / 1024 ** 2


EXPERIMENTS = {
    "Full fine-tuning": dict(method="full", r=None, lr=5e-5),
    "Faqat head (frozen)": dict(method="head", r=None, lr=1e-3),
    "LoRA r=8": dict(method="lora", r=8, lr=5e-4),
    "LoRA r=16": dict(method="lora", r=16, lr=5e-4),
}

rows, outputs = [], {}
for name, cfg in EXPERIMENTS.items():
    model = build(cfg["method"], cfg["r"])
    trainable, total = count_trainable(model), count_total(model)
    args = TrainingArguments(
        output_dir=f"out/{name.replace(' ', '_')}",
        per_device_train_batch_size=32, per_device_eval_batch_size=64,
        num_train_epochs=EPOCHS, learning_rate=cfg["lr"],
        eval_strategy="epoch", save_strategy="no", logging_strategy="no",
        report_to="none", seed=SEED, fp16=torch.cuda.is_available(),
        label_names=["labels"])
    trainer = Trainer(model=model, args=args, train_dataset=train_tok, eval_dataset=val_tok,
                      data_collator=collator, compute_metrics=compute_metrics)
    t = time.perf_counter()
    trainer.train()
    train_time = time.perf_counter() - t

    pred_out = trainer.predict(test_tok)
    logits = pred_out.predictions
    pred = logits.argmax(-1)
    save_dir = f"saved/{name.replace(' ', '_')}"
    model.save_pretrained(save_dir)

    outputs[name] = {"logits": logits, "pred": pred}
    rows.append({
        "Usul": name,
        "Test accuracy": accuracy_score(test_labels, pred),
        "Test F1 (macro)": f1_score(test_labels, pred, average="macro"),
        "Trainable params": trainable,
        "Trainable (%)": 100 * trainable / total,
        "Train vaqti (s)": train_time,
        "Saqlangan hajm (MB)": dir_size_mb(save_dir),
    })
    print(name, "tayyor")
    del model, trainer
    gc.collect()
    torch.cuda.empty_cache()

res = pd.DataFrame(rows)
print(res.round(4).to_string(index=False))

full = res[res["Usul"] == "Full fine-tuning"].iloc[0]
for _, r in res[res["Usul"].str.startswith("LoRA")].iterrows():
    print(f"{r['Usul']}: accuracy farqi {100 * (r['Test accuracy'] - full['Test accuracy']):+.2f} p.p., "
          f"F1 farqi {100 * (r['Test F1 (macro)'] - full['Test F1 (macro)']):+.2f} p.p., "
          f"hajm {full['Saqlangan hajm (MB)'] / r['Saqlangan hajm (MB)']:.0f} marta kichik")

fig, axes = plt.subplots(1, len(EXPERIMENTS), figsize=(5 * len(EXPERIMENTS), 4.5))
for ax, (name, o) in zip(axes, outputs.items()):
    cm = confusion_matrix(test_labels, o["pred"], labels=[0, 1, 2])
    ConfusionMatrixDisplay(cm, display_labels=LABELS).plot(ax=ax, colorbar=False, cmap="Blues")
    ax.set_title(name)
plt.tight_layout()
plt.show()


def top_errors(name, k=5):
    logits = outputs[name]["logits"]
    pred = outputs[name]["pred"]
    probs = torch.softmax(torch.tensor(logits), dim=-1).numpy()
    wrong = np.where(pred != test_labels)[0]
    order = wrong[np.argsort(-probs[wrong, pred[wrong]])][:k]
    print(f"\n{name}: eng ishonch bilan yanglishgan {k} ta misol")
    for i in order:
        print(f"- Matn: {test_texts[i][:120]}")
        print(f"  Haqiqiy: {LABELS[test_labels[i]]} | Bashorat: {LABELS[pred[i]]} | Ishonch: {probs[i, pred[i]]:.2f}")


best_lora = res[res["Usul"].str.startswith("LoRA")].sort_values("Test F1 (macro)").iloc[-1]["Usul"]
top_errors("Full fine-tuning")
top_errors(best_lora)
```
