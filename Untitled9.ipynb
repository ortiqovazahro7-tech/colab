```python
import time
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from scipy import linalg
from sklearn.metrics import (roc_auc_score, average_precision_score, precision_recall_curve,
                             precision_score, recall_score, f1_score)
from tensorflow import keras
from tensorflow.keras import layers

SEED, NORMAL = 42, 7
AE_EPOCHS, GAN_EPOCHS, GAN_BATCH, LATENT = 30, 60, 128, 100
keras.utils.set_random_seed(SEED)

(x_tr_all, y_tr_all), (x_te_all, y_te_all) = keras.datasets.mnist.load_data()
x_tr_all = (x_tr_all.astype("float32") / 255.0)[..., None]
x_te_all = (x_te_all.astype("float32") / 255.0)[..., None]

x_normal = x_tr_all[y_tr_all == NORMAL]
rng = np.random.default_rng(SEED)
rng.shuffle(x_normal)
n_val = int(0.1 * len(x_normal))
x_val_normal, x_train_normal = x_normal[:n_val], x_normal[n_val:]
is_anomaly = (y_te_all != NORMAL).astype(int)
print("Normal sinf:", NORMAL, "| train:", len(x_train_normal), "| val:", n_val,
      "| test:", len(x_te_all), "| anomaliya ulushi:", round(is_anomaly.mean(), 3))

def build_autoencoder():
    keras.utils.set_random_seed(SEED)
    inp = keras.Input(shape=(28, 28, 1))
    x = layers.Conv2D(32, 3, strides=2, padding="same", activation="relu")(inp)
    x = layers.Conv2D(64, 3, strides=2, padding="same", activation="relu")(x)
    x = layers.Flatten()(x)
    z = layers.Dense(16, activation="relu")(x)
    x = layers.Dense(7 * 7 * 64, activation="relu")(z)
    x = layers.Reshape((7, 7, 64))(x)
    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same", activation="relu")(x)
    x = layers.Conv2DTranspose(32, 3, strides=2, padding="same", activation="relu")(x)
    out = layers.Conv2D(1, 3, padding="same", activation="sigmoid")(x)
    m = keras.Model(inp, out)
    m.compile(optimizer=keras.optimizers.Adam(1e-3), loss="mse")
    return m

def train_autoencoder(x_train):
    m = build_autoencoder()
    es = keras.callbacks.EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
    m.fit(x_train, x_train, validation_data=(x_val_normal, x_val_normal),
          epochs=AE_EPOCHS, batch_size=128, callbacks=[es], verbose=0)
    return m

def recon_error(m, x):
    rec = m.predict(x, batch_size=512, verbose=0)
    return np.mean((x - rec) ** 2, axis=(1, 2, 3))

ae = train_autoencoder(x_train_normal)
err_test = recon_error(ae, x_te_all)
err_val = recon_error(ae, x_val_normal)

roc = roc_auc_score(is_anomaly, err_test)
pr = average_precision_score(is_anomaly, err_test)
print(f"Autoencoder: ROC-AUC = {roc:.4f} | PR-AUC = {pr:.4f}")

thr = np.percentile(err_val, 95)
pred = (err_test > thr).astype(int)
print(f"Threshold (normal validation xatolarining 95-persentili) = {thr:.5f}")
print(f"  Precision = {precision_score(is_anomaly, pred):.4f} | "
      f"Recall = {recall_score(is_anomaly, pred):.4f} | F1 = {f1_score(is_anomaly, pred):.4f}")

prec, rec, thrs = precision_recall_curve(is_anomaly, err_test)
f1s = 2 * prec[:-1] * rec[:-1] / np.maximum(prec[:-1] + rec[:-1], 1e-12)
best = int(np.argmax(f1s))
print(f"Taqqoslash uchun F1 bo'yicha eng yaxshi threshold = {thrs[best]:.5f} | F1 = {f1s[best]:.4f}")

plt.figure(figsize=(8, 5))
plt.hist(err_test[is_anomaly == 0], bins=60, alpha=0.6, density=True, label="Normal")
plt.hist(err_test[is_anomaly == 1], bins=60, alpha=0.6, density=True, label="Anomaliya")
plt.axvline(thr, color="black", linestyle="--", label="Threshold (95%)")
plt.xlabel("Reconstruction MSE")
plt.ylabel("Zichlik")
plt.legend()
plt.show()

def build_generator():
    inp = keras.Input(shape=(LATENT,))
    x = layers.Dense(7 * 7 * 256, use_bias=False)(inp)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(0.2)(x)
    x = layers.Reshape((7, 7, 256))(x)
    x = layers.Conv2DTranspose(128, 5, strides=1, padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(0.2)(x)
    x = layers.Conv2DTranspose(64, 5, strides=2, padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(0.2)(x)
    out = layers.Conv2DTranspose(1, 5, strides=2, padding="same", activation="tanh")(x)
    return keras.Model(inp, out)

def build_discriminator():
    inp = keras.Input(shape=(28, 28, 1))
    x = layers.Conv2D(64, 5, strides=2, padding="same")(inp)
    x = layers.LeakyReLU(0.2)(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Conv2D(128, 5, strides=2, padding="same")(x)
    x = layers.LeakyReLU(0.2)(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Flatten()(x)
    out = layers.Dense(1)(x)
    return keras.Model(inp, out)

keras.utils.set_random_seed(SEED)
G, D = build_generator(), build_discriminator()
g_opt = keras.optimizers.Adam(2e-4, beta_1=0.5)
d_opt = keras.optimizers.Adam(2e-4, beta_1=0.5)
bce = keras.losses.BinaryCrossentropy(from_logits=True)

@tf.function
def train_step(real):
    noise = tf.random.normal([tf.shape(real)[0], LATENT])
    with tf.GradientTape() as g_tape, tf.GradientTape() as d_tape:
        fake = G(noise, training=True)
        real_out = D(real, training=True)
        fake_out = D(fake, training=True)
        g_loss = bce(tf.ones_like(fake_out), fake_out)
        d_loss = bce(tf.ones_like(real_out) * 0.9, real_out) + bce(tf.zeros_like(fake_out), fake_out)
    g_grads = g_tape.gradient(g_loss, G.trainable_variables)
    d_grads = d_tape.gradient(d_loss, D.trainable_variables)
    g_opt.apply_gradients(zip(g_grads, G.trainable_variables))
    d_opt.apply_gradients(zip(d_grads, D.trainable_variables))
    return g_loss, d_loss

gan_data = (tf.data.Dataset.from_tensor_slices(x_normal * 2.0 - 1.0)
            .shuffle(len(x_normal), seed=SEED).batch(GAN_BATCH, drop_remainder=True)
            .prefetch(tf.data.AUTOTUNE))

t0 = time.perf_counter()
for epoch in range(1, GAN_EPOCHS + 1):
    g_l, d_l = [], []
    for batch in gan_data:
        gl, dl = train_step(batch)
        g_l.append(float(gl))
        d_l.append(float(dl))
    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch}/{GAN_EPOCHS} | G loss: {np.mean(g_l):.3f} | D loss: {np.mean(d_l):.3f}")
print("GAN o'qitish vaqti (s):", round(time.perf_counter() - t0, 1))

keras.utils.set_random_seed(SEED)
fake_imgs = G(tf.random.normal([1000, LATENT]), training=False).numpy()
fake01 = (fake_imgs + 1.0) / 2.0

fig, axes = plt.subplots(2, 10, figsize=(15, 3.4))
for i in range(10):
    axes[0, i].imshow(x_normal[i, ..., 0], cmap="gray")
    axes[1, i].imshow(fake01[i, ..., 0], cmap="gray")
    axes[0, i].axis("off")
    axes[1, i].axis("off")
axes[0, 0].set_title("Haqiqiy", fontsize=9, loc="left")
axes[1, 0].set_title("DCGAN", fontsize=9, loc="left")
plt.tight_layout()
plt.show()

inception = keras.applications.InceptionV3(include_top=False, weights="imagenet",
                                           pooling="avg", input_shape=(299, 299, 3))

def inception_features(imgs01):
    feats = []
    for i in range(0, len(imgs01), 100):
        b = tf.convert_to_tensor(imgs01[i:i + 100])
        b = tf.image.resize(tf.repeat(b, 3, axis=-1), (299, 299))
        b = keras.applications.inception_v3.preprocess_input(b * 255.0)
        feats.append(inception.predict(b, verbose=0))
    return np.concatenate(feats)

def fid_score(f1, f2):
    mu1, mu2 = f1.mean(0), f2.mean(0)
    s1, s2 = np.cov(f1, rowvar=False), np.cov(f2, rowvar=False)
    covmean = linalg.sqrtm(s1 @ s2)
    if np.iscomplexobj(covmean):
        covmean = covmean.real
    return float(np.sum((mu1 - mu2) ** 2) + np.trace(s1) + np.trace(s2) - 2 * np.trace(covmean))

f_real_a = inception_features(x_normal[:1000])
f_real_b = inception_features(x_normal[1000:2000])
f_fake = inception_features(fake01)
print(f"FID (DCGAN va haqiqiy): {fid_score(f_fake, f_real_a):.2f}")
print(f"FID (haqiqiy va haqiqiy, taqqoslash uchun): {fid_score(f_real_b, f_real_a):.2f}")

ae_aug = train_autoencoder(np.concatenate([x_train_normal, fake01.astype("float32")]))
err_aug = recon_error(ae_aug, x_te_all)
roc_aug = roc_auc_score(is_anomaly, err_aug)
pr_aug = average_precision_score(is_anomaly, err_aug)

bonus = pd.DataFrame({
    "Variant": ["Faqat haqiqiy normal", "Haqiqiy + 1000 sintetik"],
    "ROC-AUC": [roc, roc_aug],
    "PR-AUC": [pr, pr_aug],
})
print(bonus.round(4).to_string(index=False))
print(f"Farq: ROC-AUC {roc_aug - roc:+.4f} | PR-AUC {pr_aug - pr:+.4f}")
```
