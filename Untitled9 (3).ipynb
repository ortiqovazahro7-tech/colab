```python
# 2-topshiriq: Disbalansli dataset va overfitting laboratoriyasi
# 5 variant: baseline, class_weight, SMOTE, dropout+L2, early stopping+augmentation

# %% 1. Kutubxonalar (Colab'da imblearn bo'lmasa, avval bir marta ishga tushiring)
# !pip install -q imbalanced-learn
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, average_precision_score)
from imblearn.over_sampling import SMOTE

SEED, EPOCHS, BATCH = 42, 40, 256
keras.utils.set_random_seed(SEED)   # natija har safar bir xil chiqishi uchun

# %% 2. Sun'iy disbalansli dataset (positive taxminan 3%, ya'ni 5% dan kam)
X, y = make_classification(
    n_samples=20000, n_features=20, n_informative=8, n_redundant=4,
    weights=[0.97, 0.03], flip_y=0.01, class_sep=0.8, random_state=SEED)
print("Positive ulushi:", round(y.mean(), 4))

# Train 60% / validation 20% / test 20% (har birida sinf nisbati saqlanadi)
X_tr, X_tmp, y_tr, y_tmp = train_test_split(
    X, y, test_size=0.4, stratify=y, random_state=SEED)
X_va, X_te, y_va, y_te = train_test_split(
    X_tmp, y_tmp, test_size=0.5, stratify=y_tmp, random_state=SEED)

# Masshtabni train bo'yicha o'rganamiz, val va test'ga faqat qo'llaymiz
sc = StandardScaler().fit(X_tr)
X_tr, X_va, X_te = sc.transform(X_tr), sc.transform(X_va), sc.transform(X_te)


# %% 3. Model va metrikalar
def build_model(l2=0.0, dropout=0.0, noise=0.0):
    # Bir xil arxitektura, faqat regularizatsiya sozlamalari farq qiladi
    reg = keras.regularizers.l2(l2) if l2 > 0 else None
    inp = keras.Input(shape=(X_tr.shape[1],))
    h = inp
    if noise > 0:
        h = layers.GaussianNoise(noise)(h)   # augmentation: faqat o'qitishda shovqin qo'shadi
    for units in (128, 128):
        h = layers.Dense(units, activation="relu", kernel_regularizer=reg)(h)
        if dropout > 0:
            h = layers.Dropout(dropout)(h)   # tasodifiy neyronlarni o'chiradi
    out = layers.Dense(1, activation="sigmoid")(h)
    m = keras.Model(inp, out)
    m.compile(optimizer=keras.optimizers.Adam(1e-3), loss="binary_crossentropy")
    return m


def get_metrics(m, X_, y_):
    # Ehtimolni olamiz, 0.5 dan yuqorisini "positive" deb hisoblaymiz
    p = m.predict(X_, verbose=0).ravel()
    pred = (p >= 0.5).astype(int)
    return {
        "Acc": accuracy_score(y_, pred),
        "Prec": precision_score(y_, pred, zero_division=0),
        "Rec": recall_score(y_, pred, zero_division=0),
        "F1": f1_score(y_, pred, zero_division=0),
        "ROC_AUC": roc_auc_score(y_, p),
        "PR_AUC": average_precision_score(y_, p),
    }


# %% 4. Beshta variantni o'qitamiz
# class_weight: kam uchraydigan sinfga katta vazn beramiz
n, n1 = len(y_tr), y_tr.sum()
cw = {0: n / (2 * (n - n1)), 1: n / (2 * n1)}

# SMOTE: faqat TRAIN'ga qo'llaymiz (val/test'ga tegmaymiz, aks holda "leakage" bo'ladi)
X_sm, y_sm = SMOTE(random_state=SEED).fit_resample(X_tr, y_tr)

stop = keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=5, restore_best_weights=True)

variants = {
    "1) baseline":          dict(build={}, X=X_tr, y=y_tr, fit={}),
    "2) class_weight":      dict(build={}, X=X_tr, y=y_tr, fit={"class_weight": cw}),
    "3) SMOTE":             dict(build={}, X=X_sm, y=y_sm, fit={}),
    "4) dropout+L2":        dict(build={"l2": 1e-3, "dropout": 0.3}, X=X_tr, y=y_tr, fit={}),
    "5) earlystop+augment": dict(build={"noise": 0.1}, X=X_tr, y=y_tr, fit={"callbacks": [stop]}),
}

results = {}
for name, v in variants.items():
    keras.utils.set_random_seed(SEED)        # hammasi bir xil boshlang'ich holatdan
    m = build_model(**v["build"])
    m.fit(v["X"], v["y"], validation_data=(X_va, y_va),
          epochs=EPOCHS, batch_size=BATCH, verbose=0, **v["fit"])
    # Baholash har doim ASL (SMOTE'siz) train, val va test'da
    results[name] = {
        "train": get_metrics(m, X_tr, y_tr),
        "val": get_metrics(m, X_va, y_va),
        "test": get_metrics(m, X_te, y_te),
    }
    print("tayyor:", name)

# %% 5. Jadval 1: test metrikalari
test_tbl = pd.DataFrame({k: v["test"] for k, v in results.items()}).T
print("\n=== TEST metrikalari ===")
print(test_tbl.round(3).to_string())
print("Tasodifiy model PR-AUC (= positive ulushi):", round(y_te.mean(), 3))

# %% 6. Jadval 2: overfitting tahlili (train / val / test farqi)
rows = []
for k, v in results.items():
    rows.append({
        "Variant": k,
        "F1 train": v["train"]["F1"], "F1 val": v["val"]["F1"], "F1 test": v["test"]["F1"],
        "PR_AUC train": v["train"]["PR_AUC"], "PR_AUC val": v["val"]["PR_AUC"],
        "PR_AUC test": v["test"]["PR_AUC"],
        # Gap = train va test orasidagi farq. Gap qancha kichik bo'lsa, overfitting shuncha kam
        "Gap (train-test PR_AUC)": v["train"]["PR_AUC"] - v["test"]["PR_AUC"],
        "Gap (val-test PR_AUC)": v["val"]["PR_AUC"] - v["test"]["PR_AUC"],
    })
over_tbl = pd.DataFrame(rows).set_index("Variant")
print("\n=== OVERFITTING tahlili ===")
print(over_tbl.round(3).to_string())

```
