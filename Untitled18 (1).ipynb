```python
!pip install -q importlib_resources

import os
import gc
import time
import random
import numpy as np
import pandas as pd

SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)

import tensorflow as tf
tf.random.set_seed(SEED)
tf.keras.utils.set_random_seed(SEED)

gpus = tf.config.list_physical_devices("GPU")
print("TF:", tf.__version__, "| GPU:", gpus)

# tf_flowers (5 sinf, ~3670 rasm) - tez yuklanadi va transfer learning uchun qulay
import tensorflow_datasets as tfds

(ds_full,), info = tfds.load(
    "tf_flowers",
    split=["train[:85%]"],
    with_info=True,
    as_supervised=True
)
(ds_test_raw,) = tfds.load(
    "tf_flowers",
    split=["train[85%:]"],
    as_supervised=True
)

NUM_CLASSES = info.features["label"].num_classes
IMG_SIZE = 160  # Tez va aniq ishlashi uchun 160x160
BATCH_SIZE = 32

# Kichik/o'rta dataset sifatida train/val/test ga ajratish
total_tr = tf.data.experimental.cardinality(ds_full).numpy()
val_count = int(0.15 * total_tr)
train_count = total_tr - val_count

ds_full_shuf = ds_full.shuffle(total_tr, seed=SEED, reshuffle_each_iteration=False)
ds_val_raw = ds_full_shuf.take(val_count)
ds_train_raw = ds_full_shuf.skip(val_count)

print(f"Classes: {NUM_CLASSES} | Train: {train_count} | Val: {val_count} | Test: {tf.data.experimental.cardinality(ds_test_raw).numpy()}")
```

```python
from tensorflow.keras.applications import MobileNetV2, EfficientNetB0, ResNet50
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as prep_mnv2
from tensorflow.keras.applications.efficientnet import preprocess_input as prep_effb0
from tensorflow.keras.applications.resnet50 import preprocess_input as prep_res50

BACKBONE_CONFIGS = {
    "MobileNetV2": (MobileNetV2, prep_mnv2),
    "EfficientNetB0": (EfficientNetB0, prep_effb0),
    "ResNet50": (ResNet50, prep_res50),
}

aug_layer = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal", seed=SEED),
    tf.keras.layers.RandomRotation(0.1, seed=SEED),
    tf.keras.layers.RandomZoom(0.1, seed=SEED),
], name="aug")

def make_dataset(raw_ds, prep_fn, training=False):
    def _prep(img, label):
        img = tf.image.resize(img, (IMG_SIZE, IMG_SIZE))
        img = tf.cast(img, tf.float32)
        if training:
            img = aug_layer(img, training=True)
        img = prep_fn(img)
        return img, label

    ds = raw_ds.map(_prep, num_parallel_calls=tf.data.AUTOTUNE)
    if training:
        ds = ds.shuffle(1000, seed=SEED)
    return ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

def get_peak_gpu_mb():
    if not tf.config.list_physical_devices("GPU"):
        return 0.0
    try:
        mem_info = tf.config.experimental.get_memory_info("GPU:0")
        return round(mem_info["peak"] / (1024 * 1024), 1)
    except Exception:
        return 0.0

def reset_gpu_peak():
    if tf.config.list_physical_devices("GPU"):
        try:
            tf.config.experimental.reset_memory_stats("GPU:0")
        except Exception:
            pass
```

```python
def build_transfer_model(backbone_name, strategy):
    tf.keras.utils.set_random_seed(SEED)
    cls_fn, _ = BACKBONE_CONFIGS[backbone_name]
    base = cls_fn(include_top=False, weights="imagenet", input_shape=(IMG_SIZE, IMG_SIZE, 3))

    if strategy == "1_Frozen_Head":
        base.trainable = False
        lr = 1e-3
    elif strategy == "2_FineTune_Last30Pct":
        base.trainable = True
        split_idx = int(len(base.layers) * 0.7)
        for layer in base.layers[:split_idx]:
            layer.trainable = False
        lr = 1e-4
    elif strategy == "3_FineTune_All_LowLR":
        base.trainable = True
        lr = 1e-5
    else:
        raise ValueError("Noma'lum strategiya")

    inputs = tf.keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
    x = base(inputs, training=False if strategy == "1_Frozen_Head" else None)
    x = tf.keras.layers.GlobalAveragePooling2D()(x)
    x = tf.keras.layers.Dropout(0.25)(x)
    outputs = tf.keras.layers.Dense(NUM_CLASSES, activation="softmax")(x)

    model = tf.keras.Model(inputs, outputs, name=f"{backbone_name}_{strategy}")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

EPOCHS = 6  # Tez va solishtirishga qulay; GPU bo'lsa 8-10 gacha oshirish mumkin
strategies = ["1_Frozen_Head", "2_FineTune_Last30Pct", "3_FineTune_All_LowLR"]
results = []

for b_name, (_, prep_fn) in BACKBONE_CONFIGS.items():
    ds_tr = make_dataset(ds_train_raw, prep_fn, training=True)
    ds_va = make_dataset(ds_val_raw, prep_fn, training=False)
    ds_te = make_dataset(ds_test_raw, prep_fn, training=False)

    for strat in strategies:
        tf.keras.backend.clear_session()
        gc.collect()
        reset_gpu_peak()

        print(f">>> {b_name} | {strat} boshlandi...")
        model = build_transfer_model(b_name, strat)
        trainable_params = int(np.sum([tf.keras.backend.count_params(w) for w in model.trainable_weights]))
        total_params = int(np.sum([tf.keras.backend.count_params(w) for w in model.weights]))

        t0 = time.perf_counter()
        model.fit(ds_tr, validation_data=ds_va, epochs=EPOCHS, verbose=0)
        fit_time = time.perf_counter() - t0

        test_loss, test_acc = model.evaluate(ds_te, verbose=0)
        peak_mem_mb = get_peak_gpu_mb()

        # Inference vaqti (test dataset bo'yicha o'rtacha ms/rasm)
        t_inf0 = time.perf_counter()
        n_imgs = 0
        for batch_x, _ in ds_te:
            _ = model(batch_x, training=False)
            n_imgs += batch_x.shape[0]
        inf_ms_per_img = ((time.perf_counter() - t_inf0) / max(n_imgs, 1)) * 1000.0

        row = {
            "Backbone": b_name,
            "Strategy": strat,
            "Test_Acc": round(float(test_acc), 4),
            "Trainable_Params": trainable_params,
            "Trainable_Pct": round(100.0 * trainable_params / max(total_params, 1), 2),
            "Peak_GPU_MB": peak_mem_mb,
            "Fit_Time_s": round(fit_time, 2),
            "Inference_ms_img": round(inf_ms_per_img, 3),
        }
        results.append(row)
        print(f"    Tugadi: Acc={row['Test_Acc']:.4f} | Trainable={trainable_params:,} | PeakGPU={peak_mem_mb} MB | Inf={inf_ms_per_img:.2f} ms")

df_task5 = pd.DataFrame(results)
display(df_task5)
df_task5.to_csv("task5_transfer_learning_9exp.csv", index=False)

print("\nEng yaxshi 3 ta kombinatsiya (Test_Acc bo'yicha):")
display(df_task5.sort_values(by="Test_Acc", ascending=False).head(3))
```
